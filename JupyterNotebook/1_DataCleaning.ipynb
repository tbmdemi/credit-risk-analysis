```python
import pandas as pd
```

```python
df = pd.read_csv("../data/raw/UCI_Credit_Card.csv")
```

```python
df.info()
```

```python
# mv means missing values
mv = df.isnull()
print(mv.sum())
```

```python
df.drop_duplicates()
```

```python
# In ra cac gia tri co trong du lieu
print("Education:", df["EDUCATION"].unique())
print("Marriage:", df["MARRIAGE"].unique())
print(f"Age: {df['AGE'].min()} - {df['AGE'].max()}")
```

```python
# Gom nhom EDUCATION (5,6) ve 4 (others)
df["EDUCATION"] = df["EDUCATION"].replace([5, 6], 4)

# Gom nhom MARRIAGE (0) ve 3
df['MARRIAGE'] = df['MARRIAGE'].replace(0, 3)

df.rename(columns={
    'default.payment.next.month': "IS_DEFAULT",
    'PAY_0': 'PAY_1'
}, inplace=True)
```

```python
# Truoc khi mapping thì lưu file raw duoi dạng parquet
df.to_parquet(path="../data/processed/UCI_Credit_Card.parquet")
```

```python
# Chuẩn hóa các giá trị trong dataset
df['SEX'] = df['SEX'].map({
    1: 'Nam',
    2: 'Nữ'
})

df['MARRIAGE'] = df['MARRIAGE'].map({
    1: 'Đã kết hôn',
    2: 'Độc thân',
    3: 'Khác'
})

df['EDUCATION'] = df['EDUCATION'].map({
    1: 'Sau đại học',
    2: 'Đại học',
    3: 'Phổ thông',
    4: 'Khác'
})
```

```python
df.describe()
```

```python
df.to_csv("../data/processed/UCI_Credit_Card_Cleaned.csv", index=False)
```

```python
df.to_parquet("../data/processed/UCI_Credit_Card_Cleaned.parquet", engine= "pyarrow")
```
